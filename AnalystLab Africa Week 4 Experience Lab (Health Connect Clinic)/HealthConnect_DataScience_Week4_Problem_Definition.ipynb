# Part 6 — Handling Appointment Outcomes and Cancellations

The appointment_outcome variable contains three outcomes:

- Attended
- No-Show

Since the defined machine learning problem is to predict whether a patient will `fail to attend a scheduled appointment without cancelling`, these outcomes should not all be treated as the same class.

## 6.1 Proposed Outcome Handling

The proposed approach is:
| Appointment Outcome | Modelling Treatment                               |
| ------------------- | ------------------------------------------------- |
| `No-Show`           | Retain as the positive class (`1`)                |
| `Attended`          | Retain as the negative class (`0`)                |
| `Cancelled`         | Exclude from the binary no-show modelling dataset |

**Justification**

`Cancelled is a different outcome from a No-Show`.

**A patient who cancels an appointment has communicated that they will not attend, while a no-show occurs when the patient does not attend the appointment without prior cancellation. Since the machine learning problem focuses specifically on predicting no-shows, including cancelled appointments in either the Attended or No-Show class could introduce ambiguity into the target.**

Therefore, the proposed modelling dataset will contain only:

- No-Show = 1
- Attended = 0

Cancelled appointments will be removed only when creating the future binary modelling dataset.

## 1. Assumptions

The proposed approach assumes that:

- Each record represents a valid scheduled appointment.
- The `appointment_outcome` accurately reflects whether an appointment was attended, resulted in a no-show, or was cancelled.
- `previous_appointments` and `previous_no_shows` represent information available before the current appointment.
- Variables selected as model features will be available at the time a prediction is made.
- Multiple records for the same `patient_id` represent different legitimate appointments.

## 2. Limitations

The current approach has the following limitations:

- The dataset is fictional and anonymized.
- The dataset contains limited missing values in some variables.
- The project does not currently provide information beyond the available dataset variables.
- `appointment_outcome` includes cancelled appointments, requiring a decision when constructing the binary modelling dataset.
- Some variables require further confirmation regarding when they become available before being used for prediction.
`age_group` and some appointment-related date variables may contain information that is derived or redundant.

## 3. Risks

The proposed approach may be affected by the following risks:

- **Data leakage:** Using information unavailable at prediction time could produce unrealistically high model performance.
- **Repeated patient records**: Appointments from the same patient appearing in both training and testing datasets could affect evaluation.
- **Target definition risk**: An inappropriate treatment of cancelled appointments could make the no-show target ambiguous.
-** Missing-data risk:** Poor handling of missing values could affect model performance.
- **Feature redundancy:** Closely related or derived variables may introduce unnecessary duplication of information.
- **Evaluation risk:** Relying on a single performance metric could provide an incomplete view of model performance.

## 4. Dependencies

The future development of the model depends on:

- Availability of the HealthConnect appointment dataset.
- Availability of the Data Dictionary for correct variable interpretation.
- A confirmed target definition and cancellation-handling strategy.
- Availability of relevant information at the time predictions are made.
- Appropriate data preparation and feature engineering.
- A suitable data-splitting strategy.
- Access to the required Python libraries and machine learning environment.
